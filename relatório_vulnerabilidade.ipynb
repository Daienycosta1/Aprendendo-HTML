````python
import google.generativeai as genai
import json
from datetime import datetime
````

````python
GOOGLE_API_KEY = "https://aistudio.google.com/app/apikey"
genai.configure(api_key=GOOGLE_API_KEY)
````

````python
model = genai.GenerativeModel('gemini-2.0-flash-exp')
generation_config = {
    "temperature": 0.6,
    "top_p": 0.95,
    "max_output_tokens": 500
}
````

````python
prompt = """Você é um especialista em segurança de software. Analise o seguinte código Python para um sistema de consultas a uma base de livros sobre IA generativa e gere um relatório de vulnerabilidades:
```python
def consultar_livros(tema):
    query = f"SELECT  FROM livros WHERE tema = '{tema}'"
    return executar_query(query)

Inclua:

Identificação de vulnerabilidades (ex.: injeção de SQL)
Sugestões de correções
Impacto potencial
Use um tom profissional para GitHub."""
````

````python
textos = []
for i in range(5):
  try:
    resultado = model.generate_content(prompt, generation_config=generation_config)
    textos.append(resultado.text.strip())
    print(f"? Geração {i+1} concluída")
  except Exception as e:
    textos.append(f"Erro na geração: {str(e)}")
    print(f"? Erro na geração {i+1}: {str(e)}")
````

````python
saidas_validas = []
for i, texto in enumerate(textos):
  if len(texto.split()) > 10 and "vulnerabilidade" in texto.lower():  # Critérios flexíveis
    saidas_validas.append({"id": i+1, "texto": texto})
````

````python
if saidas_validas:
  melhor_relatorio = saidas_validas[0]  # Escolhe o primeiro válido
  with open('relatorio_vulnerabilidades.md', 'w', encoding='utf-8') as f:
    f.write(f"# Relatório de Vulnerabilidades - LivrosAI\n")
    f.write(f"# Gerado em: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}\n\n")
    f.write(melhor_relatorio['texto'])
    print(f"\n?? MELHOR RELATÓRIO (ID: {melhor_relatorio['id']}):")
    print("=" * 60)
    if len(melhor_relatorio['texto']) > 500:
      print(melhor_relatorio['texto'][:500] + "...")
    else:
      print(f"\n?? Arquivo salvo: relatorio_vulnerabilidades.md")
else:
  print("? Nenhuma saída atendeu aos critérios de consistência.")
````
